```python
import gymnasium as gym
from stable_baselines3 import DQN
from stable_baselines3.common.evaluation import evaluate_policy
import os

# --- 1. Ortamı Kurma ve Klasörler Oluşturma ---
log_dir = "logs/"
model_dir = "models/DQN_vs_DDQN/"
os.makedirs(log_dir, exist_ok=True)
os.makedirs(model_dir, exist_ok=True)

# LunarLander-v3 ortamını oluşturuyoruz.
env = gym.make("LunarLander-v3")

# --- 2. Modelleri Tanımlama ve Eğitme ---

# Eğitim için ortak parametreler
TOTAL_TIMESTEPS = 150000

# a) Standart DQN (Baseline)
# double_q parametresi varsayılan olarak False'tur, bu yüzden bu standart DQN'dir.
print("--- Standart DQN Eğitimi Başlatılıyor ---")
model_standart = DQN(
    "MlpPolicy",
    env,
    verbose=0,
    tensorboard_log=log_dir
)
model_standart.learn(total_timesteps=TOTAL_TIMESTEPS, tb_log_name="DQN_Standard_Lunar")
model_standart.save(os.path.join(model_dir, "dqn_standard_lunar_model"))
print("Standart DQN Eğitimi Tamamlandı.")

# b) Double DQN
# Double DQN'i aktive etmek için tek yapmamız gereken 'double_q=True' parametresini eklemek.
print("\n--- Double DQN Eğitimi Başlatılıyor ---")
model_double = DQN(
    "MlpPolicy",
    env,
    double_q=True,  # Double Q-Learning mekanizmasını aktive et
    verbose=0,
    tensorboard_log=log_dir
)
model_double.learn(total_timesteps=TOTAL_TIMESTEPS, tb_log_name="DQN_Double_Lunar")
model_double.save(os.path.join(model_dir, "dqn_double_lunar_model"))
print("Double DQN Eğitimi Tamamlandı.")

# --- 3. Modelleri Karşılaştırmalı Olarak Değerlendirme ---
print("\n--- Modeller Değerlendiriliyor ---")
n_eval = 20  # Değerlendirme için bölüm sayısı

mean_reward_standart, std_reward_standart = evaluate_policy(model_standart, env, n_eval_episodes=n_eval)
mean_reward_double, std_reward_double = evaluate_policy(model_double, env, n_eval_episodes=n_eval)

# Sonuçları bir tablo formatında yazdıralım (Ablation Table)
print("\n--- SONUÇLAR (DQN vs DDQN) ---")
print(f"{'Model Türü':<20} | {'Ortalama Ödül':<20} | {'Standart Sapma':<20}")
print("-" * 65)
print(f"{'Standart DQN':<20} | {mean_reward_standart:<20.2f} | {std_reward_standart:<20.2f}")
print(f"{'Double DQN':<20} | {mean_reward_double:<20.2f} | {std_reward_double:<20.2f}")
print("-" * 65)

print("\nTensorBoard loglarını incelemek için terminalde 'tensorboard --logdir logs' komutunu çalıştırın.")
print("Bu komutla iki modelin öğrenme eğrilerini ('ep_rew_mean') karşılaştırarak hangisinin daha hızlı ve kararlı öğrendiğini analiz edebilirsiniz.")

env.close()
```
