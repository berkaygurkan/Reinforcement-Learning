```python
import torch
import numpy as np
import random
import gymnasium as gym
import stable_baselines3 as sb3

print(f"PyTorch Version: {torch.__version__}")
print(f"NumPy Version: {np.__version__}")
print(f"Gymnasium Version: {gym.__version__}")
print(f"Stable-Baselines3 Version: {sb3.__version__}")


# --- Tekrarlanabilirlik için Seed Ayarlama Fonksiyonu ---
# Bir RL deneyinde rastgelelik kaynakları şunlar olabilir:
# 1. Ağırlıkların başlangıç değerleri (Policy ve Value network'leri)
# 2. Ortamın (environment) başlangıç durumu
# 3. Ortamın `step` fonksiyonundaki rastgelelikler (eğer varsa)
# 4. Aksiyon seçimindeki rastgelelik (örneğin, stokastik bir politikadan örneklem alırken)
# 5. Veri zenginleştirme (Data Augmentation) veya Replay Buffer'dan veri çekerken
# Bu yüzden, kullandığımız tüm kütüphanelerin seed'lerini ayarlamak kritiktir.

def set_seed(seed_value=42):
    """Deneylerin tekrarlanabilirliğini sağlamak için seed ayarlar."""
    torch.manual_seed(seed_value)
    torch.cuda.manual_seed_all(seed_value) # Eğer CUDA kullanılıyorsa
    np.random.seed(seed_value)
    random.seed(seed_value)
    # Gymnasium ortamları için de seed ayarı gerekebilir
    # env.action_space.seed(seed_value)
    
    # PyTorch'un deterministik algoritmalar kullanmasını sağla
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
    print(f"All random seeds set to: {seed_value}")

# Fonksiyonu çağıralım
set_seed(42)

# --- Cihaz (Device) Kontrolü ---
# M1/M2/M3 çipli Mac'lerde PyTorch, GPU hızlandırması için "mps" (Metal Performance Shaders) kullanır.
# Bu, NVIDIA GPU'lardaki "cuda"nın karşılığıdır.

if torch.backends.mps.is_available():
    device = torch.device("mps")
    print("\nMPS (Apple Silicon GPU) is available!")
elif torch.cuda.is_available():
    device = torch.device("cuda")
    print("\nCUDA (NVIDIA GPU) is available!")
else:
    device = torch.device("cpu")
    print("\nNo GPU available, using CPU.")

print(f"Selected device: {device}")
```
